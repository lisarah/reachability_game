```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import MDP_algorithms.mdp_state_action as mdp
import MDP_algorithms.value_iteration as vi
import potential_games as game

# --- 1. Configuration ---
Rows, Columns = 4, 10
T = 15
player_num = 3
BR_iter = 10
MCs = 10  # 10 cases per entropy level
entropy_levels = [0.75, 0.85, 0.95]

all_data = []

# --- 2. Simulation Loop ---
for p_val in entropy_levels:
    print(f"Running entropy level: {p_val}")
    for tr in range(MCs):
        # Fresh transition template for this entropy
        P_template = mdp.transitions(Rows, Columns, p=p_val, with_stay=True) 
        reachable_set = mdp.reachable_set(P_template)
        S, _, A = P_template.shape

        initial_inds = np.random.choice(np.arange(0, 11), size=player_num, replace=False) 
        target_inds = np.random.choice(np.arange(30, 40), size=player_num, replace=False)
        targ_raw_inds = sorted(list(target_inds))   
        start_raw_inds = sorted(list(initial_inds), reverse=True)

        # Initialize Policies
        Ps, pols, rhos = [], [], []
        for p in range(player_num):
            Ps.append(P_template.copy())
            Ps[p][:, targ_raw_inds[p], 4] = 0.
            Ps[p][targ_raw_inds[p], targ_raw_inds[p], 4] = 1.
            c = np.zeros(S); c[targ_raw_inds[p]] = 1.
            Vk, new_pi = vi.value_iteration(Ps[p], c, T)
            pols.append(new_pi)
            rhos.append(mdp.occupancy(pols[p], Ps[p], start_raw_inds[p]))

        # Best Response Loop
        curr_p = -1
        for ind in range(BR_iter):
            curr_p = (curr_p + 1) % player_num
            
            # Execute your exact algorithm
            V, no_col_rate, W = game.multiplicative_values(curr_p, pols, targ_raw_inds, Ps, rhos, reachable_set)
            Vp, new_pi = vi.net_value_iteration(Ps[curr_p], W)
            new_rho = mdp.occupancy(new_pi, Ps[curr_p], start_raw_inds[curr_p]) 
            
            # Record current policies
            pols[curr_p] = new_pi  
            rhos[curr_p] = new_rho 

            # Reach Numerator Calculation
            current_reach_probs = [r[targ_raw_inds[i], T] for i, r in enumerate(rhos)]
            reach_numerator = np.prod(current_reach_probs)
            
            all_data.append({
                'Iteration': ind,
                'Potential': V,
                'Collision Likelihood': no_col_rate,
                'Reach Numerator': reach_numerator,
                'Entropy': p_val,
                'Trial': tr
            })
```
